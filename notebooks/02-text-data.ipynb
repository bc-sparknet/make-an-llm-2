## Step 1: Splitting text into tokens

```python
import re

text = "Hello, world. Is this-- a test?"
tokens = re.split(r'([,.:;?_!"()\']|--|\s)', text)
tokens = [t.strip() for t in tokens if t.strip()]
print(tokens)
# ['Hello', ',', 'world', '.', 'Is', 'this', '--', 'a', 'test', '?']
```

## Step 2: Mapping tokens to IDs

```python
class SimpleTokenizer:
    def __init__(self, vocab):
        self.str_to_int = vocab                                  # "fox" -> 12
        self.int_to_str = {i: s for s, i in vocab.items()}      # 12 -> "fox"

    def encode(self, text):
        tokens = [t.strip() for t in re.split(r'([,.:;?_!"()\']|--|\s)', text) if t.strip()]
        tokens = [t if t in self.str_to_int else "<|unk|>" for t in tokens]
        return [self.str_to_int[t] for t in tokens]

    def decode(self, ids):
        text = " ".join(self.int_to_str[i] for i in ids)
        return re.sub(r'\s+([,.?!"()\'])', r'\1', text)  # re-attach punctuation
```

## Step 3: Byte-pair encoding

```python
import tiktoken

tokenizer = tiktoken.get_encoding("gpt2")
text = "Good morning! <|endoftext|> The robot tasted a glorptastic sandwich."
ids = tokenizer.encode(text, allowed_special={"<|endoftext|>"})

print(len(ids))                          # more tokens than words: rare words get split
print(50256 in ids)                      # True: <|endoftext|> has ID 50256
print([tokenizer.decode([i]) for i in ids])  # see where the word boundaries fall
print(tokenizer.decode(ids) == text)     # True: encoding is lossless
```

## Step 4: Creating input–target pairs

```python
import tiktoken
import torch
from torch.utils.data import Dataset, DataLoader

class NextTokenDataset(Dataset):
    def __init__(self, text, tokenizer, context_length, stride):
        ids = tokenizer.encode(text, allowed_special={"<|endoftext|>"})
        self.inputs, self.targets = [], []
        for start in range(0, len(ids) - context_length, stride):
            window = ids[start : start + context_length + 1]
            self.inputs.append(torch.tensor(window[:-1]))   # all but the last
            self.targets.append(torch.tensor(window[1:]))   # all but the first

    def __len__(self):
        return len(self.inputs)

    def __getitem__(self, i):
        return self.inputs[i], self.targets[i]


def create_dataloader(text, batch_size=8, max_length=256, stride=256,
                      shuffle=True, drop_last=True):
    dataset = NextTokenDataset(text, tiktoken.get_encoding("gpt2"),
                               context_length=max_length, stride=stride)
    return DataLoader(dataset, batch_size=batch_size,
                      shuffle=shuffle, drop_last=drop_last)
```

```python
with open("short_story.txt", encoding="utf-8") as f:
    raw_text = f.read()

loader = create_dataloader(raw_text, batch_size=8, max_length=256, stride=256)

inputs, targets = next(iter(loader))
print(inputs.shape)   # torch.Size([8, 256]): 8 windows of 256 token IDs
```

## Step 5: Token embeddings

```python
vocab_size = 50257
emb_dim = 256

token_embedding = torch.nn.Embedding(vocab_size, emb_dim)
token_vectors = token_embedding(inputs)
print(token_vectors.shape)  # torch.Size([8, 256, 256]): batch × tokens × emb_dim
```

## Step 6: Adding position information

```python
context_length = 256
pos_embedding = torch.nn.Embedding(context_length, emb_dim)

positions = torch.arange(context_length)          # 0, 1, 2, … 255
input_vectors = token_vectors + pos_embedding(positions)
print(input_vectors.shape)  # torch.Size([8, 256, 256])
```
